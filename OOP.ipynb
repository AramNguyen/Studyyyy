```python
class student:
    def __init__(self, age, type_):
        self.age = age
        self.type_ = type_
    def onboard(self):
        print(f"Bạn Phú, {self.age} đang được GPA {self.type_}")
Phu = student(20, "A+")
Phu.onboard()

class place(student):
    def __init__(self, name, home, year):
        self.name = name
        self.home = home
        self.year = year
    def board2(self):
        print(f"Bạn {self.name} đã sống ở {self.home} từ những năm {self.year}")
Phu = place("Phu", "Long Xuyen", "2000")
Phu.board2()
```
