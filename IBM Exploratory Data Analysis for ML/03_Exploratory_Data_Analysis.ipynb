# SAMPLING FROM DATAFRAME : 

### SAMPLING : 

- Take a small piece from a bigger dataframe to check the characteristics of the dataframe


### CODE : 

`sample = data.sample(n = 5, replace = False)`

`print(sample.iloc[:,-3:])`

- For a futher explanation : 

    + n = 5 is the number of columns we take out for sample

    + replace = False is to make sure those 5 columns are not the same 

    + **In the real-world EDA, we common use pd.head(5) or pd.tail(5) to check the structure of the dataframe**
