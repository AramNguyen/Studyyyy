# SAMPLING FROM DATAFRAME : 

### SAMPLING : 

- Take a small piece from a bigger dataframe to check the characteristics of the dataframe


### CODE : 

`sample = data.sample(n = 5, replace = False)`

`print(sample.iloc[:,-3:])`

- For a futher explanation : 

    + n = 5 is the number of columns we take out for sample

    + replace = False is to make sure those 5 columns are not the same 

    + **In the real-world EDA, we common use pd.head(5) or pd.tail(5) to check the structure of the dataframe**

# VARIABLE SELEECTION : BACKGROUND 

- **Variable selection** involves choosing the set of features to include in model 

    (chon ra mot tap hop cac dac trung de dua vao model)

- Variable must often be transformed before they can be included in models

- In addition to log and polynominal transformations, this can be involve :

    + Encoding : converting non-numeric features to numeric features

    + Scaling : converting the scale numeric data so thay can compare 

**The appropriate method of scaling or encoding depends on the type of features**


## FEATURES ENCODING : TYPE OF FEATURES

- Encoding is often applied to categorical features, that take non-numeric values. 

- Two primary types : 

    + Nominal : categorical variables take values in unordered catergories

        (e.g. Red, Blue, True, False)

    + Ordinal : catergorical variables take values in odered catergories

        (e.g. High, Moderate, Low, Cold, Warm, Hot)



## FEATURE ENCODING : APPROACHES 

- There are several common approaches to encoding variables :

    + Binary encoding : converts variables to either 1 or 0 and is suitable for variables that take two possible values (e.g. True, False, Male, Female, Married, Not Married)

    + One - hot encoding : converts variables that take multiple values into binary (0 or 1) variables, one for each category. This create several new variables

    + Ordinal encoding : involves converting ordered categories to numerical values, usually by creating one variable that take integer equal to the number of catogories (e.g. 0, 1, 2, 3, 4,...)

        (example : low --> 0, medium --> 1, high --> 2, extremly high --> 3)

    


## FEATURES SCALING : BACKGROUND

- Features scaling involves adjusting a variable's scale

This allow comparison of variables with different scales

- Different continuos (numeric) features often have different scales

- We must convert the variables more carefull, not to let something too high in a high scale compare to to low in a low scale (for example: 3.5M sec compare to 1-10 time surgery)



## FEATURES SCALING : APPROACHES

- There are many approaches to scaling features : 

    + Standard scaling : convert features to standard normal variables

        (by subtracting the mean and dividing by the standard error)

    + More strongly: **Min - Max Scaling** : converts variables to continous variables in the (0, 1) interval by mapping minimum values to 0 and maximum values to 1
