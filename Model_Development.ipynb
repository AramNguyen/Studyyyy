### LINER REGRESSION AND MULTIPLE LINEAR 

---


### LINER REGRESSION 

- Will refer to one independent variables to make a prediction

independent variable x1 --> **Linear Regression** --> prediction

### MULTIPLE INDEPENDENT 

-  Will refer to multiple independent variables to make a prediction

multiple independent variables x1, x2, xn... --> **Multiple Linear Regression** --> prediction

### SIMPLE LINEAR REGRESSION

- The predictor (independent) variable - x

- The target (dependent) variablem -y

`y = b0 + b1x`

- b0 : the intercept (Điểm chặn)

- b1 : the slope (Độ dốc)



### Import lineear_model for scikit-learn

`from sklearn.linear_model import LinearRegression`


### Create a Linear Regression Object
