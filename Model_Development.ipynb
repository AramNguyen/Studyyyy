### R-SQUARED (R^2)

+ The coefficient of Determination or R-Squared (R^2)

+ Is the measure to determine how close the data is to be fitted regression line

+ R^2 : the percentage of variation of the target variable (Y) that is explained by the linear model 

**Think about as comparing a regression model to a simple model i.e the mean of the data point**

**MSE của đường hồi quy / MSE trung bình của data**

|Đường màu xanh là đường hồi quy|
|:--|
|Chấm đen là điểm dữ liệu|
|Blue squared là MSE của đường hồi quy|
|Đường màu đỏ là đường của dữ liệu trung bình bảng|
|Red squared là MSE của đường trung bình|
|**Ta nhận thấy cái blue squared nhỏ hơn các red squared**|

Because of this is a good fit, so the value will close to zero, R^2 = (1 - MSE Regression / MSE average)
