## Introduction to Simple Linear Regression

### How simple linear regression works

- The target value represented as the response value --> Y_hat

- The slope (do doc) theta one

### Introduction to Multiple Linear Regression

- Extension of simple linear regression model

- Use two or more independent variables to estimate a dependent variable

- Theta 0 is a bias while theta 1 to theta n is a weight of each x1 to xn respectively

- The more features to depend on, the graph will be defined by a line if it is one feature and a plane if there are two features, the more features -> the more dimensionality will be displayed

### Compare Multiple linear regression with simple linear regression

- Multiple linear regression better model than simple linear regression 

- Too many variables can cause overfitting - model will learn all the data by accident --> it's performance in the new data will less accuracy

- To improve prediction, convert categorical independent variables to numerical variables --> Can understand this by convert the categorical variables into numerical variables by encoding (ordinal or one-hot)


### Applications of Multiple Linear Regression 


- Used in education to predict outcomes and explain relationships. For examples : do revision time, test anxiety, attendance,...

- Used to predict the impact of changes in "what if" scenarios 

### "What if" scenario pitfalls

- Inaccurate findings : 

    + Considering impossible scenarios 

    + Extrapolating scenarios

    + Model might depend on a group of correlated or colinear variables

### Correlation pitfalls 

- Correlation variables are no longer "independent variables"

- If variable correlated with another feature, "what if" scenario not possible 


## Solution to correlation pitfalls 

- Remove redundant variables

- Select variables which are : 

    + Most understood

    + Controllable 

    + Most correlated with target 

### Prediction with multiple linear regression 

- Multiple linear regression assigns relative importance to features

### Fitting a hyperplane 

- Simple linear regression : Regression equation defines a line 

- Multiple linear regression using two features : Solution describes a plane

- Beyond two dimensions : Solution describes a hyperplane

### Model Error

- Average of all residual errors indicates how poorly model predicts the actual values

    -> This information is called the mean squared error or MSE

### Least squares solution 

- The best model for the dataset is the one with the least squared error

### Estimating parameters for multiple linear regression 

- Ordinary least squares and an optimization approach are the most common methods

- Ordinary least squares estimate coefficients by minimizing MSE

- This approach uses linear algebra to calculate optimal theta

## Training a Logistic Regression Model 

### Logistic regression training 

- Identify parameters that map input features to target outcomes 

- Objective: Predict classes with minimal error 

- Find theta/parameters that minimizes cost function



**Pipelines**

- Choose a set of parameters or theta

- Predict probability that class = 1

- Calculate prediction error (cost function)

- Update theta 0 to reduce prediction error

- Repeat until: 

    + Reach small log-loss value or 

    + Targeted number of iterations

### Optimal logistic regression 

# Optimal Logistic Regression Notes

## 1. Concepts
- **Preliminary Model**: A binary classifier formed by passing a linear combination through a sigmoid function ($\sigma(\mathbf{X}\boldsymbol{\theta})$).
- **Optimization Requirement**: Finding optimal parameters $\boldsymbol{\theta} = [\theta_0, \theta_1, \dots]^T$ using an evaluation metric (Binary Cross-Entropy / Log Loss) via Gradient Descent.

## 2. Key Equations
- **Logit / Linear prediction**: 
  $$\hat{y} = \mathbf{X}\boldsymbol{\theta}$$
- **Sigmoid activation**: 
  $$\hat{p} = \sigma(\hat{y}) = \frac{1}{1 + \exp(-\hat{y})}$$
- **Binary Decision rule**:
  $$\hat{C} = 1 \text{ if } \hat{p} \ge 0.5 \text{ else } 0$$
- **Cost Function (Binary Cross-Entropy / Log Loss)**:
  $$J(\boldsymbol{\theta}) = -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} \log(\hat{p}^{(i)}) + (1 - y^{(i)}) \log(1 - \hat{p}^{(i)}) \right]$$

- VNESE: 

Dựng xong công thức Sigmoid giống như bạn mới lắp ráp xong cái xe đạp (preliminary model). Bạn không thể phóng nhanh ngay được mà phải qua bước cân chỉnh xích, tra dầu, chỉnh phanh (optimization) dựa trên cảm giác lái thử (loss metric) thì xe mới đạt trạng thái chạy ngon nhất (optimal).

- Cost function or log-loss needs to be minimized 

- Measures how well $(\hat{p}_i)$ matches $y_i$

### Log-Loss Intuition

- **Confident and correct**: Predicted probability of class 1 is high and correct $\implies$ log-loss is small.
  - *Example*: $y = 1, \hat{p} = 0.99 \implies -\log(0.99) \approx 0.01$

- **Confident and incorrect**: Predicted probability of class 0 is high and incorrect $\implies$ log-loss is large.
  - *Example*: $y = 1, \hat{p} = 0.01 \implies -\log(0.01) \approx 4.60$

- One more thing to remind: Log-loss is a value that represents both the accuracy and the cofident of the model

### Minimizing cost function with gradient descent 

**How ?**

- Stop training when log-loss is statisfactory

- Use gradient descent

**What is gradient descent ?**

### What is Gradient Descent?

- **Iterative optimization**: An iterative approach to finding the minimum of a cost/loss function.
- **Parameter update**: Adjusts parameter values ($\theta$) in the opposite direction of the log-loss derivative ($\nabla J(\theta)$).
  - Formula: $\theta := \theta - \alpha \frac{\partial J}{\partial \theta}$
- **Learning rate ($\alpha$)**: Determines the step size taken towards the minimum at each iteration to avoid overshooting or slow convergence.

### Gradient descent on a surface 

**Goal:** Change parameter values and find path to optimal parameters to minimize the cost function


- **Objective**: Iteratively adjust parameters ($\theta_1, \theta_2$) along the gradient path to reach the global minimum of the cost function (Log Loss).
- **Optimization Path**: The algorithm steps down the curved error surface in the direction of steepest descent until convergence.
- **Global Minimum**: The lowest point of the surface corresponds to the optimal set of parameters $(\theta_1^*, \theta_2^*)$.

### Gradient Descent 

- Calculated over the entire data set 

- Large data set = Slow descent

- Converge less likely as steps too big to notice minima 

- Gradient can be approximated using a random subset

### Stochastic gradient descent (SGD)


- Variation of the gradient descent 

- Uses a random data subset and scales well 

- Likely to overlook local to find global minima

- Converges quickly toward a global minimum



- Convergence can be improved by: 

    + Decreasing learning rate 

    + Gradually increasing sample size 

### Tổng kết — Logistic Regression & Tối ưu hóa

- **Mục tiêu Logistic Regression:** Dự đoán lớp với **sai số tối thiểu**

- **Training:** Tìm bộ tham số $\theta$ sao cho **Cost Function nhỏ nhất**

- **Optimization:** Dùng thuật toán tối ưu để tìm **bộ tham số tốt nhất**

- **Log-loss:** Là **Cost Function** cần được tối thiểu hóa

- **Log-loss:** Thưởng cho dự đoán **đúng + tự tin**, phạt nặng dự đoán **sai + quá tự tin**

- **Gradient Descent:** Thuật toán lặp để tìm **giá trị nhỏ nhất của một hàm**

- **SGD (Stochastic Gradient Descent):** Biến thể của Gradient Descent, dùng **một tập con dữ liệu ngẫu nhiên**, giúp xử lý dataset lớn tốt hơn

```python
"""
REGRESSION CHEATSHEET - PRACTICE SCRIPT
=======================================

Script này minh họa các kiến thức trong cheatsheet:
1. train_test_split
2. StandardScaler
3. Simple Linear Regression
4. Multiple Linear Regression
5. Polynomial Regression
6. Logistic Regression
7. log_loss
8. mean_absolute_error (MAE)
9. mean_squared_error (MSE)
10. root mean squared error (RMSE)
11. r2_score

Mục tiêu:
- Đọc code từ trên xuống dưới.
- Chạy thử từng model.
- Nhìn vào phần comment để hiểu mỗi dòng đang làm gì.
"""

import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import (
    log_loss,
    mean_absolute_error,
    mean_squared_error,
    r2_score,
)


# ============================================================
# 0. TẠO DATASET
# ============================================================
# Dataset regression:
# X1 = số giờ học
# X2 = số giờ ngủ
# y  = điểm số
#
# Đây là dữ liệu giả lập để mình có thể chạy toàn bộ ví dụ
# mà không cần tải dataset từ Internet.

X_reg = np.array([
    [1, 5],
    [2, 5],
    [2, 6],
    [3, 6],
    [3, 7],
    [4, 6],
    [4, 7],
    [5, 7],
    [5, 8],
    [6, 7],
    [6, 8],
    [7, 8],
    [7, 9],
    [8, 8],
    [8, 9],
    [9, 9],
], dtype=float)

# Điểm số có quan hệ gần tuyến tính với số giờ học + số giờ ngủ.
y_reg = np.array([
    45, 49, 53, 57,
    60, 63, 67, 71,
    75, 78, 81, 85,
    88, 91, 94, 97
], dtype=float)


# ============================================================
# 1. train_test_split
# ============================================================
# Chia dữ liệu thành:
# - Training set: dùng để train model
# - Testing set: dùng để kiểm tra model trên dữ liệu chưa thấy
#
# test_size=0.2 nghĩa là 20% dữ liệu dành cho testing.
# random_state=42 giúp lần chạy sau vẫn chia giống lần này.

X_train, X_test, y_train, y_test = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)

print("=" * 70)
print("1. train_test_split")
print("=" * 70)
print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)


# ============================================================
# 2. StandardScaler
# ============================================================
# StandardScaler chuẩn hóa feature:
#
# x_scaled = (x - mean) / standard_deviation
#
# fit_transform(X_train):
#   - fit    = học mean và standard deviation từ TRAINING data
#   - transform = dùng chúng để chuẩn hóa training data
#
# transform(X_test):
#   - KHÔNG fit lại test data
#   - dùng mean/std đã học từ training data
#
# Đây là cách làm đúng để tránh data leakage.

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("\n2. StandardScaler")
print("First scaled training sample:", X_train_scaled[0])


# ============================================================
# 3. SIMPLE LINEAR REGRESSION
# ============================================================
# Simple Linear Regression:
#
# y = b0 + b1*x
#
# Chỉ dùng 1 independent variable.
# Ở đây ta chỉ dùng feature X1 = số giờ học.

X_simple = X_reg[:, [0]]  # Lấy riêng cột số giờ học.

X_simple_train, X_simple_test, y_simple_train, y_simple_test = train_test_split(
    X_simple,
    y_reg,
    test_size=0.2,
    random_state=42
)

# Tạo model.
simple_model = LinearRegression()

# Train model:
# model sẽ tìm b0 và b1 phù hợp với training data.
simple_model.fit(X_simple_train, y_simple_train)

# Dự đoán trên test set.
y_simple_pred = simple_model.predict(X_simple_test)

print("\n" + "=" * 70)
print("3. Simple Linear Regression")
print("=" * 70)
print("Equation: y = b0 + b1*x")
print("b0 (intercept) =", simple_model.intercept_)
print("b1 (coefficient) =", simple_model.coef_[0])
print("Predictions =", y_simple_pred)


# ============================================================
# 4. MULTIPLE LINEAR REGRESSION
# ============================================================
# Multiple Linear Regression dùng nhiều independent variables:
#
# y = b0 + b1*x1 + b2*x2 + ...
#
# Ở đây:
# x1 = số giờ học
# x2 = số giờ ngủ

multiple_model = LinearRegression()

# Train model với cả 2 feature.
multiple_model.fit(X_train_scaled, y_train)

# Dự đoán trên test data.
y_multiple_pred = multiple_model.predict(X_test_scaled)

print("\n" + "=" * 70)
print("4. Multiple Linear Regression")
print("=" * 70)
print("Equation: y = b0 + b1*x1 + b2*x2 + ...")
print("Coefficients =", multiple_model.coef_)
print("Intercept =", multiple_model.intercept_)


# ============================================================
# 5. POLYNOMIAL REGRESSION
# ============================================================
# Polynomial Regression dùng các feature dạng:
#
# x, x^2, x^3, ...
#
# Ví dụ degree=2:
#
# y = b0 + b1*x + b2*x^2
#
# Quan trọng:
# Polynomial Regression vẫn dùng LinearRegression ở phía sau.
# PolynomialFeatures chỉ tạo thêm các feature x^2, x^3,...

X_poly_base = X_simple

X_poly_train, X_poly_test, y_poly_train, y_poly_test = train_test_split(
    X_poly_base,
    y_reg,
    test_size=0.2,
    random_state=42
)

# Tạo feature polynomial bậc 2.
poly = PolynomialFeatures(degree=2)

# Biến:
# [x] -> [1, x, x^2]
X_poly_train_transformed = poly.fit_transform(X_poly_train)
X_poly_test_transformed = poly.transform(X_poly_test)

# Dùng LinearRegression trên các feature polynomial.
poly_model = LinearRegression()
poly_model.fit(X_poly_train_transformed, y_poly_train)

y_poly_pred = poly_model.predict(X_poly_test_transformed)

print("\n" + "=" * 70)
print("5. Polynomial Regression")
print("=" * 70)
print("Degree =", 2)
print("Features after transformation:")
print(poly.get_feature_names_out(["hours_studied"]))
print("Predictions =", y_poly_pred)


# ============================================================
# 6. REGRESSION METRICS
# ============================================================
# Các metric dưới đây dùng để đánh giá regression model.
#
# MAE:
#   Mean Absolute Error
#   = trung bình |y_true - y_pred|
#   → càng nhỏ càng tốt
#
# MSE:
#   Mean Squared Error
#   = trung bình (y_true - y_pred)^2
#   → càng nhỏ càng tốt
#
# RMSE:
#   Root Mean Squared Error
#   = sqrt(MSE)
#   → càng nhỏ càng tốt
#
# R^2:
#   Cho biết model giải thích được bao nhiêu biến thiên của target.
#   Thường càng gần 1 càng tốt.

mae = mean_absolute_error(y_test, y_multiple_pred)
mse = mean_squared_error(y_test, y_multiple_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_multiple_pred)

print("\n" + "=" * 70)
print("6. Regression Metrics")
print("=" * 70)
print("MAE  =", mae)
print("MSE  =", mse)
print("RMSE =", rmse)
print("R^2  =", r2)

# Một số phiên bản sklearn mới có hàm root_mean_squared_error().
# Nếu muốn dùng trực tiếp, có thể viết:
#
# from sklearn.metrics import root_mean_squared_error
# rmse = root_mean_squared_error(y_test, y_multiple_pred)
#
# Nhưng sqrt(MSE) phía trên giúp thấy rõ bản chất RMSE.


# ============================================================
# 7. LOGISTIC REGRESSION
# ============================================================
# Logistic Regression dùng cho CLASSIFICATION.
#
# Ví dụ:
# 0 = Không đậu
# 1 = Đậu
#
# Model dự đoán probability thuộc class 1.
#
# Logistic Regression có thể viết:
#
# log(p / (1-p)) = b0 + b1*x1 + b2*x2 + ...
#
# Trong sklearn:
# LogisticRegression()
# model.fit(X, y)
# model.predict(X)
# model.predict_proba(X)

# Tạo target classification:
# Nếu điểm >= 70 -> 1 (đậu)
# Nếu điểm < 70  -> 0 (không đậu)
y_class = (y_reg >= 70).astype(int)

X_class_train, X_class_test, y_class_train, y_class_test = train_test_split(
    X_reg,
    y_class,
    test_size=0.25,
    random_state=42,
    stratify=y_class
)

# Logistic Regression thường hoạt động tốt hơn khi feature được scale.
class_scaler = StandardScaler()

X_class_train_scaled = class_scaler.fit_transform(X_class_train)
X_class_test_scaled = class_scaler.transform(X_class_test)

# Tạo Logistic Regression model.
logistic_model = LogisticRegression()

# Train model.
logistic_model.fit(X_class_train_scaled, y_class_train)

# predict() -> trả về class 0 hoặc 1.
y_class_pred = logistic_model.predict(X_class_test_scaled)

# predict_proba() -> trả về probability của từng class.
y_class_proba = logistic_model.predict_proba(X_class_test_scaled)

# Lấy probability của class 1.
y_class_proba_positive = y_class_proba[:, 1]

print("\n" + "=" * 70)
print("7. Logistic Regression")
print("=" * 70)
print("Classes =", logistic_model.classes_)
print("Predicted classes =", y_class_pred)
print("Probability of class 1 =", y_class_proba_positive)


# ============================================================
# 8. LOG-LOSS
# ============================================================
# Log-loss đánh giá chất lượng probability prediction
# của classification model.
#
# Ý tưởng:
# - Dự đoán đúng + tự tin -> loss nhỏ
# - Dự đoán sai + quá tự tin -> loss lớn
#
# Vì vậy:
# LOG-LOSS CÀNG NHỎ -> CÀNG TỐT

loss = log_loss(y_class_test, y_class_proba)

print("\n" + "=" * 70)
print("8. Log-loss")
print("=" * 70)
print("Log-loss =", loss)


# ============================================================
# 9. TÓM TẮT
# ============================================================
print("\n" + "=" * 70)
print("CHEATSHEET SUMMARY")
print("=" * 70)

print("""
Regression:
- Simple Linear Regression  -> 1 feature
- Multiple Linear Regression -> nhiều features
- Polynomial Regression -> thêm x^2, x^3, ...
- MAE -> sai số tuyệt đối trung bình
- MSE -> bình phương sai số trung bình
- RMSE -> sqrt(MSE)
- R^2 -> mức độ model giải thích biến thiên của target

Classification:
- Logistic Regression -> dự đoán class / probability
- log_loss -> đo chất lượng probability prediction

Preprocessing / Evaluation:
- train_test_split -> chia train/test
- StandardScaler -> chuẩn hóa features
""")

print("Script finished successfully.")
```
