## Introduction to Simple Linear Regression 

- Linear Regression

    + Models linear relationship between continuos target variable and explanatory features

    + Can be use to predict a continuos value

- Simple Linear Regression

    + Single independent variable estimates dependent variable 

    + CO2 emission predicted using EngineSize variable



### How simple linear regression works

- You can determine a "best-fit" line through the data

- The target value represented as the response value --> Y_hat

- The slope (do doc) theta one

- Theta zero or theta one are called the coefficients og the linear regression model

    (Selected by the linear regression algorithms to determine a best-fit line)

- We can use two formulas to calculate coefficients theta 0 and theta 1 

    -> It required that we calculate the means, y-bar (trung binh y) and x-bar (trung binh x) of independent and dependent variables

### Pros and cons of OLS regression 


- OLS regression method is helpful because it's easy to understand and interpret

- The method doesn't require any tuning, solution just a calculation 

- However, accuracy can be greatly reduced by outcomes

    * A linear model may be too far too simplistic to capture complexity, such as nonlinear relationship in the data
