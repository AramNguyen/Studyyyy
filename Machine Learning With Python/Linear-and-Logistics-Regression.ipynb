## Introduction to Simple Linear Regression 

- Linear Regression

    + Models linear relationship between continuos target variable and explanatory features

    + Can be use to predict a continuos value

- Simple Linear Regression

    + Single independent variable estimates dependent variable 

    + CO2 emission predicted using EngineSize variable



### How simple linear regression works

- You can determine a "best-fit" line through the data

- The target value represented as the response value --> Y_hat

- The slope (do doc) theta one

- Theta zero or theta one are called the coefficients og the linear regression model

    (Selected by the linear regression algorithms to determine a best-fit line)

- We can use two formulas to calculate coefficients theta 0 and theta 1 

    -> It required that we calculate the means, y-bar (trung binh y) and x-bar (trung binh x) of independent and dependent variables

### Pros and cons of OLS regression 


- OLS regression method is helpful because it's easy to understand and interpret

- The method doesn't require any tuning, solution just a calculation 

- However, accuracy can be greatly reduced by outcomes

    * A linear model may be too far too simplistic to capture complexity, such as nonlinear relationship in the data

### Introduction to Multiple Linear Regression

- Extension of simple linear regression model

- Use two or more independent variables to estimate a dependent variable

- Theta 0 is a bias while theta 1 to theta n is a weight of each x1 to xn respectively

- The more features to depend on, the graph will be defined by a line if it is one feature and a plane if there are two features, the more features -> the more dimensionality will be displayed

### Compare Multiple linear regression with simple linear regression

- Multiple linear regression better model than simple linear regression 

- Too many variables can cause overfitting - model will learn all the data by accident --> it's performance in the new data will less accuracy

- To improve prediction, convert categorical independent variables to numerical variables --> Can understand this by convert the categorical variables into numerical variables by encoding (ordinal or one-hot)


### Applications of Multiple Linear Regression 


- Used in education to predict outcomes and explain relationships. For examples : do revision time, test anxiety, attendance,...

- Used to predict the impact of changes in "what if" scenarios 

### "What if" scenario pitfalls

- Inaccurate findings : 

    + Considering impossible scenarios 

    + Extrapolating scenarios

    + Model might depend on a group of correlated or colinear variables

### Correlation pitfalls 

- Correlation variables are no longer "independent variables"

- If variable correlated with another feature, "what if" scenario not possible 


## Solution to correlation pitfalls 

- Remove redundant variables

- Select variables which are : 

    + Most understood

    + Controllable 

    + Most correlated with target 

### Prediction with multiple linear regression 

- Multiple linear regression assigns relative importance to features

### Fitting a hyperplane 

- Simple linear regression : Regression equation defines a line 

- Multiple linear regression using two features : Solution describes a plane

- Beyond two dimensions : Solution describes a hyperplane

### Model Error

- Average of all residual errors indicates how poorly model predicts the actual values

    -> This information is called the mean squared error or MSE

### Least squares solution 

- The best model for the dataset is the one with the least squared error

### Estimating parameters for multiple linear regression 

- Ordinary least squares and an optimization approach are the most common methods

- Ordinary least squares estimate coefficients by minimizing MSE

- This approach uses linear algebra to calculate optimal theta 

**---**

- Another option is to use an optimization algorithms to find the best parameters

- Gradient descent starts optimization with random values for each coefficient

- Repeat until:

# Optimal Logistic Regression Notes

## 1. Concepts
- **Preliminary Model**: A binary classifier formed by passing a linear combination through a sigmoid function ($\sigma(\mathbf{X}\boldsymbol{\theta})$).
- **Optimization Requirement**: Finding optimal parameters $\boldsymbol{\theta} = [\theta_0, \theta_1, \dots]^T$ using an evaluation metric (Binary Cross-Entropy / Log Loss) via Gradient Descent.

## 2. Key Equations
- **Logit / Linear prediction**: 
  $$\hat{y} = \mathbf{X}\boldsymbol{\theta}$$
- **Sigmoid activation**: 
  $$\hat{p} = \sigma(\hat{y}) = \frac{1}{1 + \exp(-\hat{y})}$$
- **Binary Decision rule**:
  $$\hat{C} = 1 \text{ if } \hat{p} \ge 0.5 \text{ else } 0$$
- **Cost Function (Binary Cross-Entropy / Log Loss)**:
  $$J(\boldsymbol{\theta}) = -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} \log(\hat{p}^{(i)}) + (1 - y^{(i)}) \log(1 - \hat{p}^{(i)}) \right]$$

- VNESE: 

Dựng xong công thức Sigmoid giống như bạn mới lắp ráp xong cái xe đạp (preliminary model). Bạn không thể phóng nhanh ngay được mà phải qua bước cân chỉnh xích, tra dầu, chỉnh phanh (optimization) dựa trên cảm giác lái thử (loss metric) thì xe mới đạt trạng thái chạy ngon nhất (optimal).

- Cost function or log-loss needs to be minimized 

- Measures how well $(\hat{p}_i)$ matches $y_i$

### Log-Loss Intuition

- **Confident and correct**: Predicted probability of class 1 is high and correct $\implies$ log-loss is small.
  - *Example*: $y = 1, \hat{p} = 0.99 \implies -\log(0.99) \approx 0.01$

- **Confident and incorrect**: Predicted probability of class 0 is high and incorrect $\implies$ log-loss is large.
  - *Example*: $y = 1, \hat{p} = 0.01 \implies -\log(0.01) \approx 4.60$

- One more thing to remind: Log-loss is a value that represents both the accuracy and the cofident of the model

### Minimizing cost function with gradient descent 

**How ?**

- Stop training when log-loss is statisfactory

- Use gradient descent

**What is gradient descent ?**

### What is Gradient Descent?

- **Iterative optimization**: An iterative approach to finding the minimum of a cost/loss function.
- **Parameter update**: Adjusts parameter values ($\theta$) in the opposite direction of the log-loss derivative ($\nabla J(\theta)$).
  - Formula: $\theta := \theta - \alpha \frac{\partial J}{\partial \theta}$
- **Learning rate ($\alpha$)**: Determines the step size taken towards the minimum at each iteration to avoid overshooting or slow convergence.

### Gradient descent on a surface 

**Goal:** Change parameter values and find path to optimal parameters to minimize the cost function


- **Objective**: Iteratively adjust parameters ($\theta_1, \theta_2$) along the gradient path to reach the global minimum of the cost function (Log Loss).
- **Optimization Path**: The algorithm steps down the curved error surface in the direction of steepest descent until convergence.
- **Global Minimum**: The lowest point of the surface corresponds to the optimal set of parameters $(\theta_1^*, \theta_2^*)$.

### Gradient Descent 

- Calculated over the entire data set 

- Large data set = Slow descent

- Converge less likely as steps too big to notice minima 

- Gradient can be approximated using a random subset
