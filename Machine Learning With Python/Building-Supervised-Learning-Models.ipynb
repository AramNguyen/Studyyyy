## CLASSIFICATION 

### What is classification ?

- Supervised Machine Learning method 

- Uses fully trained models to predict lables on new data

- Lables from a categorical variable with discrete values

- Understands data in context when answering a question 

- Ensures accuracy in predictions 

- Model adjusts the data to fit the algorithms and classifies it accordingly

### Applications of Classification

- Problems expressed as associations between feature and target variables 

- Used to build apps for: 

    + Email filtering 

    + Speech-to-text 

    + Handwriting recognition 

    + Biometric identification 

    + Document classification
    

### Classìication Algorithms

### CAN CLASSIFY MULTIPLE DISTINCT LABLES 

### Multiclass Prediction 

- Classification algorithms used as components for multiclass classifiers

- Strategies : 

    + One-versus-all 

    + One-versus-one

## One-versus-all Stategy: 

- **Binary classifier:** One for each class lable 

- Assigned a single lable that defines targer class 

- **Task:** Binary prediction for every data point for a one-versus-the-rest classifier

- **K-classes** = K binary classifier

## One-versus-one strategy: 

- Instead of letting one versus all, we separate them to match only two object for each classifier

# DECISION TREES 

- Each internal node corresponds to a test 

    (Mỗi nút bên trong tương ứng với một thử nghiệm)

- Each branch corresponds to the result of the test 

    (Mỗi nhánh tương ứng với một kết quả của thử nghiệm)

- Each terminal, or leaf node, assigns its data to a class 

    (Mỗi nút đầu cuối hoặc nút lá gắn giá trị dữ liệu của nó cho một lớp)

## How to build a decision tree

- Consider features of a data set 

- Example in medical study: Age, sex, blood pressure, and cholesterol

- Use training part of data set to build decision tree 

- Use decision tree to predict class of unknown patient 



## Training a decision tree


- Start with a seed node and labled training data 

        (Bắt đầu với một nút hạt giống và dữ liệu được đào tạogắn nhãn)

- Find the feature that best spilts the data 

        (Tìm feature tốt nhất thành các lớp được gắng nhẵn trước của nó)

- Each split partitions the node's input

        (Mỗi phân vùng phân chia như vậy phân vùng dữ liệu đầu vào của nút và mỗi phân vùng được truyền dọc theo nhánh của nó đến một nút mới)

- Repeat the process for each new node 

The tree grows until all nodes contain a single class each, or you run out of features to select, or a pre-select stopping criterion is met
