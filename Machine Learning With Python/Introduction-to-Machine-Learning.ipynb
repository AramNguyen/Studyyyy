### Machine Learning Techniques 

* Classification predicts class or category of a case 

* Regression/estimation predicts continuos values 

* Clustering groups similar data

* Association finds items or events that co-occur 


- Anomaly detection discovers abnormal and unusual cases 

- Sequence mining predicts next event

- Dimension reduction reduces size of data 

- Recommendation systems associate people's preferences

### Classification and regression 

- Classification : 

    * Predicts class membership of new observations 

- Regression : 

    * Predicts continuous numerical values and input data 


### Clustering 

- Unsupervised machine learning technique 

- Used to group similar data points or objects

# A Day In The Life Of A Machine Learning Engineer

- Objectives : 

    * Describe the importance and requirements of each process in the lifecycle of a machine learning model 

    * Name the processes that are more-time consuming than others

- Problem Definition: 

    * As a beauty product customer, I would like to receive recommendations for other products based on my history purchase so that i will able to address my skin care need and improve the overall health of my skin

    For example: 
    
                -> Purchase history : the product A + the product B 

                -> Recommended : the product C

- Data Collection : 

    - **User data :** 

        * Demograph, purchase history, etc
        
    - **Product data :**

        * Inventory, Ingredients, popularity, etc

    - **Other data:** 

        * Saved products, like products, search history, etc
    

- Data preparation : 

    + Errors

    + Different Formatting

    + Missing Data 
    

So we must : 

+ Clean data to remove irrelevant data 

+ Remove extreme values 

+ Missing values should be removed or replaced

+ Each data column should be in the proper format

+ Create new features 

+ Exploratory Data Analysis 

    *(validate the data based on information - do some correlation analysis)*

+ Identify how to split up the data into training and test set


### Model Development
