## Data scientist vs AI engineer 

- Data scientists have always used AI models to do their analysis

- Data Scientist : Data Storyteller 

- Works with structured data which contains a hundreds to hundreds of thousands of observations (100 - 100K observations) - require cleaning and preprocessing 

    * Descriptive : 

        * EDA

        * Clusting

    * Predictive :
        
        * Regression model
        
        * Classification model

- AI Engineer : AI-System Builder - works much with unstructured data such as text, images, videos, audio files, etc - require between billions to trillions of tokens 

    * Prescriptive :

        * Decision optimization

        * Recommendation Engines

    * Generative : 

        * Intelligent assistants 

        * Creation of chatbots

- The next area of difference, which is the underlying models:

    * The data science toolbox consists of hundreds of different models and different algorithms

        + That mean the scope of these individual models is a lot more narrow

        + Take smaller size in terms of the number of parameters - take less compute power to train and do inference - take less time to train (anywhere between seconds to hours)

    * The AI engineer's toolbox is a lot less cluttered - which only contains foundation model 

        + That mean the scope is called more wide 

        + Due to the sophistication of these models, they are a lot larger in size, often billions of parameters - require a lot more compute to train - require a lot more time to train (weeks and months)

- The underlying processes and techniques that are used to develop solutions with these also differ : 

    * Data Scientist : start off with a use case and then from that use case, you pick the right data --> After the data is prepared --> use to train and validate a model using techniques such as features engineering, cross-validation or hyper-parameters tuning 

        (This model then is deployed at some endpoint for example: in the cloud, to do real-time prediction and inference)

    * AI Engineer : Also start off with a use case, but then we can skip directly to working with a pre-trained model 

        (Interact with these foundation model via natural language instruction to prompt them to do various task --> this is known as prompt engineering)

        Framework : 
        
        * Chaining different prompt together

        * Parameter-efficient fine-tuning - PEFT - on domain-specific data

        * Retrieval-augmented generation - RAG - to ground answer in truth

        * AI Agent

        * Embed the AI

    ### Feature Scaling: Chuẩn hóa dữ liệu với StandardScaler

Chuẩn hóa các đặc trưng (features) về phân phối chuẩn có $\mu = 0$ và $\sigma = 1$:

```python
from sklearn import preprocessing

# Chuẩn hóa dữ liệu X bằng StandardScaler
X = preprocessing.StandardScaler().fit(X).transform(X)
```

> **Cách viết rút gọn tương đương:**
> ```python
> X = preprocessing.StandardScaler().fit_transform(X)
> ```

* **`StandardScaler()`**: Khởi tạo bộ chuẩn hóa theo công thức $z = \frac{x - \mu}{s}$.
* **`.fit(X)`**: Tính giá trị trung bình ($\mu$) và độ lệch chuẩn ($\sigma$) từ tập dữ liệu $X$.
* **`.transform(X)`**: Áp dụng phép biến đổi chuẩn hóa lên $X$ dựa trên các tham số đã tính.
* **`.fit_transform(X)`**: Gộp cả hai bước tính toán và biến đổi trong cùng một phương thức để tối ưu hiệu năng.

### Phân loại với Support Vector Classification (SVC)

Khởi tạo mô hình phân loại SVM với hàm nhân RBF (Radial Basis Function mặc định):

```python
from sklearn import svm

# Khởi tạo mô hình phân loại SVC với các siêu tham số tùy chỉnh
clf = svm.SVC(gamma=0.001, C=100.)
```

* **`from sklearn import svm`**: Import module SVM từ thư viện Scikit-learn.
* **`svm.SVC(...)`**: Lớp mô hình **Support Vector Classification** dùng cho bài toán phân loại (classification).
* **`C=100.`** *(Regularization parameter)*: Tham số điều chuẩn (penalty parameter).
  * `C` càng lớn: Phạt nặng các điểm phân loại sai $\rightarrow$ ranh giới quyết định (decision boundary) chặt chẽ hơn, dễ dẫn đến **Overfitting**.
  * `C` càng nhỏ: Cho phép bỏ qua một số lỗi để tạo lề phân cách rộng hơn $\rightarrow$ khái quát hóa tốt hơn, nhưng nếu quá nhỏ dễ dẫn đến **Underfitting**.
* **`gamma=0.001`** *(Kernel coefficient)*: Hệ số xác định tầm ảnh hưởng của một điểm dữ liệu đơn lẻ đối với ranh giới quyết định (áp dụng cho kernel `rbf`, `poly`, `sigmoid`).
  * `gamma` nhỏ ($0.001$): Bán kính ảnh hưởng rộng, đường biên quyết định mượt mà và phẳng hơn.
  * `gamma` lớn: Bán kính ảnh hưởng hẹp, mô hình uốn lượn sát quanh từng điểm dữ liệu, dễ dẫn đến **Overfitting**.
* **`clf`**: Viết tắt của *Classifier*, biến lưu đối tượng mô hình chuẩn bị đem đi huấn luyện (`clf.fit(X_train, y_train)`).
