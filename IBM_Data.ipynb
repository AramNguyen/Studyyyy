**Concepts of the Python Database API**

- Connection object (Đối tượng kết nối) - Cursor object (Đối tượng truy vấn) -\

++ CO: Kết nối với cơ sở dữ liệu và quản lí các giao dịch

++ CO: Chạy truy vấn (Database queries)


**Cursor Method**

cursor(): Return a new corsor object using the connection

commit(): Commit (Lưu thay đổi) any pending transaction to the database

rollback(): Causes the database to ​roll back to the start of any pending transaction (Đưa database về thời điểm ban đầu của bất kì giao dịch đang chờ nào)

close(): Close database

```python
import sqlite3

# 1. Create a connection object 
# Using ':memory:' creates a temporary RAM database that disappears when the script finishes
connection = sqlite3.connect(':memory:')

# 2. Create a cursor object
cursor = connection.cursor()

# --- Setup: Creating a table and inserting data so we have something to select ---
cursor.execute('''CREATE TABLE mytable (id INTEGER, field TEXT)''')
cursor.execute('''INSERT INTO mytable VALUES (1, 'Data Science')''')
cursor.execute('''INSERT INTO mytable VALUES (2, 'Machine Learning')''')
connection.commit() # Save the changes
# ---------------------------------------------------------------------------------

# 3. Run queries
cursor.execute('SELECT * FROM mytable')
result = cursor.fetchall()

# Print the result to verify it works
print("Query Results:")
for row in result:
    print(row)

# 4. Free resources
cursor.close()
connection.close()
```

|**TYPE**|**USE**|
|:--|:--|
|Target Variable (Dependent Variable)| Definition of the column you trying to predict|
|Independent Variable| "Other colums" used as inputs to make the prediction|
|Categorical Variable| Describes the format of the data itself, not its role in the predition model|
|Index Variable| A row number or identifier --> keep the datasheet organized, but it has no mathematical predictive value| 

|**SYNTAX - FUNCTION**|**USE**|
|:--|:--|
|connect()|Opens a connection to the specified database file - Establishes the initial link between your Python script and the database|
|execute() or **cursor.execute("SELECT * FROM table")**| Define the SQL query to run|
|fetchall() or fetchone()| (Kéo các kết quả dữ liệu về) Fetches query result after the query has been executed|
|to_csv()| Export the data to a CSV file|

```python
import pandas as pd
import matplotlib.pyplot as plt
url = '/mnt/e/Toán - Tin TCTA/Data Import/winequality-red-selected-columns.csv'
df = pd.read_csv(url, header = 0)
#print(df)

#df.describe()

#df.dtypes

df.describe(include = "all")

df.info()

x = df["fixed acidity"]
y = df["pH"]
plt.scatter (x,y)

plt.title("The relationship between fixed acidity and pH")
plt.xlabel("fixed acidity")
plt.ylabel("pH")

plt.savefig("My_scatter_plot.png")

z = df["chlorides"]
o = df["residual sugar"]
plt.figure()
plt.plot(z, o, color="red", label = "Line")
plt.title("Line Graph")
plt.legend()
plt.show()
```
